### 02) WAP to find Maximum and Minimum K elements in a given tuple.

```python

def max_min_k_elements(t, k):
    if k > len(t):
        return "Error: k is larger than the tuple size"
    
    sorted_tuple = sorted(t)
    
    min_k_elements = sorted_tuple[:k]
    max_k_elements = sorted_tuple[-k:]
    
    return min_k_elements, max_k_elements


t2 = (12, 3, 5, 7, 19, 1, 15)

k = 3
min_elements, max_elements = max_min_k_elements(t2, k)
print(f"Minimum {k} elements: {min_elements}")
print(f"Maximum {k} elements: {max_elements}")
```

### 05) WAP to find tuples with all positive elements from the given list of tuples.

```python
def positive_t(list_of_tuples):
    positive_tuples = []
    
    for t in list_of_tuples:

        if all(x > 0 for x in t):
            positive_tuples.append(t)
    
    return positive_tuples


list_of_tuples = [(1, 2, 3), (-1, 2, 3), (4, 5), (0, 2, 3), (7, 8, 9)]
result = positive_t(list_of_tuples)
print("Tuples with all positive elements:", result)

    
```
